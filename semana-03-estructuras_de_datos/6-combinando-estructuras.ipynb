# Introducción

Ahora que conoces las estructuras de datos, veremos ejemplos de uso en que se combinan para tomar ventaja de sus propiedades.

## Sets y Dicts con tuplas

En el siguiente ejemplo, usaremos un `set` para almacenar coordenadas. Podemos aprovechar de que las `tuplas` son estructuras de datos inmutables y hasheables, por lo que pueden ser almacenados en un set.

```python
class PuntosDeInteres:
    # Clase que recuerda los puntos de interes visitados en un mapa

    def __init__(self):
        # Set con puntos de interes
        self.visitados = set()

    def agregar_punto_de_interes(self, latitud, longitud):
        # Al ser la tupla hasheable, podemos agregarla en un set
        # y no se almacenarán pares repetidos
        self.visitados.add((longitud, latitud))
```

De manera similar, podemos combinar diccionarios y tuplas para además almacenar información sobre los puntos de interés

```python
class PuntosDeInteresConInfo:
    # Clase que recuerda los puntos de interes visitados en un mapa

    def __init__(self):
        # Set con puntos de interes
        self.visitados_con_info = {}

    def agregar_punto_de_interes(self, latitud, longitud, info):
        # Al ser la tupla hasheable, podemos agregarla en un set
        # y no se almacenarán pares repetidos
        self.visitados_con_info[(longitud, latitud)] = info
```

## Diccionarios con sets

Imagina que tienes un catálago de productos en una tienda online. Cada producto tiene `tags` que ayudan a clasificar objetos para facilitar la búsqueda. Podemos usar diccionarios y sets para facilitar la búsqueda usando tags de la siguiente forma.

```python
from collections import defaultdict



class Catalogo:
    def __init__(self):
        # Un diccionario cuya key es un tag (string)
        # Y los valores son sets con productos (strings)
        # Utilizamos in defaultdict para que por defecto si un valor no existe,
        # se crea un set vacio cuando se accede al valor
        self.productos_por_tag = defaultdict(set)

    def agregar_tag_a_producto(self, producto: str, tag: str):
        # Agregamos el producto al set de productos con el tag dado. 
        # Esto indica que el producto dato pertenece al tag.
        # Como usamos un default dict, no es necesario crear el
        # set si el tag aun no existe
        self.productos_por_tag[tag].add(producto)

    def buscar_productos_con_tags(self, tags: list[str]) -> set[str]:
        # Dada una lista de tags, retornamos un set con todos los productos
        # que tengan esta combinación de tags

        # Obtenemos el set de productos del primer tag
        productos = self.productos_por_tag[tags[0]]

        # Si hay mas tags, procedemos a buscar la intersección
        if len(tags) > 1:
            for tag in tags[1:]:
                productos = productos.intersection(self.productos_por_tag[tag])

        return productos
```

```python
productos = {
    "fruta": ["piña", "pera", "manzana", "platano", "limon"],
    "verdura": ["apio", "tomate", "lechuga", "pimenton"],
    "verde": ["pera", "manzana", "apio", "lechuga", "pimenton"],
    "amarillo": ["platano", "piña", "limon", "pimenton"],
    "rojo": ["manzana", "tomate", "pimenton"],
}

# Crear catalogo y agregar productos
catalogo = Catalogo()
for tag, lista_productos in productos.items():
    for producto in lista_productos:
        catalogo.agregar_tag_a_producto(producto, tag)
```

```python
# Consultar productos por tag
print(catalogo.buscar_productos_con_tags(["verde"]))
print(catalogo.buscar_productos_con_tags(["rojo"]))
print(catalogo.buscar_productos_con_tags(["fruta"]))
print(catalogo.buscar_productos_con_tags(["fruta", "amarillo"]))
print(catalogo.buscar_productos_con_tags(["verdura", "rojo", "verde"]))
```
