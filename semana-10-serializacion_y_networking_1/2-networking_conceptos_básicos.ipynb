<a id="networking"></a>
## Conceptos básicos de redes de datos (*networking*)

Las redes permiten que dos procesos se comuniquen, tanto si se ejecutan en la misma máquina como si se encuentran en dispositivos distintos.

<a id="comunicacion"></a>
### ¿Cómo se comunican los computadores?

Necesitamos:

1. Un **medio** para transmitir señales, como Ethernet, fibra óptica, Wi-Fi o redes móviles.
2. Una forma de identificar la interfaz de destino, como una **dirección IP**.
3. Una forma de identificar el proceso de destino: un **puerto**.
4. Un **protocolo de transporte**, como TCP o UDP.
5. Un **protocolo de aplicación**, que define el significado y formato de los mensajes, como HTTP, DNS o SMTP.

<a id="direccionamiento"></a>
### Direcciones IP, DNS y puertos

#### Dirección IP

Una dirección IP identifica una **interfaz de red** dentro de un ámbito. Una dirección privada debe ser única dentro de su red local; una dirección pública enrutable debe ser única en Internet. Un dispositivo puede tener varias interfaces y direcciones.

- **IPv4:** utiliza 32 bits, por ejemplo `192.0.2.10`.
- **IPv6:** utiliza 128 bits, por ejemplo `2001:db8:85a3::8a2e:370:7334`.

IPv4 dispone de $2^{32}$ combinaciones e IPv6 de $2^{128}$, aunque no todas se asignan directamente a dispositivos.

#### DNS

**DNS** significa *Domain Name System*. Traduce nombres como `cursos.canvas.uc.cl` a una o más direcciones IP mediante un sistema distribuido y jerárquico.

#### Puertos

Los puertos identifican puntos de comunicación de la capa de transporte. Tienen 16 bits y valores entre `0` y `65535`. Su asignación depende del protocolo y la dirección local; TCP y UDP pueden usar el mismo número de puerto simultáneamente.

| Puerto | Protocolo o servicio |
|---:|---|
| 21 | FTP, conexión de control |
| 22 | SSH |
| 25 | SMTP |
| 53 | DNS |
| 80 | HTTP |
| 110 | POP3 |
| 143 | IMAP |
| 443 | HTTPS |

En FTP, el puerto 21 se usa para control; el 20 puede utilizarse para datos en modo activo.

![Puertos convencionales de servicios de red](./img/ports.svg)

Para contactar un servicio normalmente necesitamos un host, un protocolo de transporte y un puerto.

<a id="transporte"></a>
## Protocolos de transporte

Estudiaremos principalmente **TCP** y **UDP**. La información se encapsula por capas: TCP usa **segmentos**, UDP **datagramas** y la capa IP los transporta en **paquetes IP**. Los puertos pertenecen a TCP o UDP; las direcciones pertenecen a IP.

<a id="tcp"></a>
### TCP (*Transmission Control Protocol*)

TCP ofrece a la aplicación un **flujo ordenado y confiable de bytes**. No conserva los límites de los mensajes de la aplicación.

TCP establece una conexión mediante un *handshake*, utiliza números de secuencia, ACK acumulativos, retransmisiones y mecanismos de control de flujo y congestión. Durante el *handshake* pueden negociarse opciones como MSS, escalamiento de ventana y SACK; no se negocia directamente la memoria de la aplicación.

HTTP/1.1, HTTP/2 y SMTP suelen usar TCP. HTTP/3 es una excepción: utiliza QUIC sobre UDP.

<a id="udp"></a>
### UDP (*User Datagram Protocol*)

UDP envía **datagramas independientes** sin establecer una conexión previa. Conserva sus límites, pero no garantiza entrega, orden ni retransmisión.

Se utiliza, por ejemplo, en DNS, telefonía y video en tiempo real, juegos y como base de QUIC. Esto no significa que todo audio o video utilice UDP: también existen soluciones sobre TCP y HTTP.

<a id="encapsulamiento"></a>
## Encapsulamiento

Cada capa ofrece servicios a la capa superior y agrega información de control. OSI y TCP/IP son modelos distintos, aunque tienen correspondencias aproximadas:

![Comparación aproximada de los modelos OSI y TCP/IP](./img/osi.png)

Nuestros programas operan en la capa de aplicación y utilizan servicios del sistema operativo, como sockets TCP o UDP. Los datos se encapsulan al bajar por la pila y se desencapsulan en el destino.

<a id="cliente-servidor"></a>
## Arquitectura cliente-servidor

Un proceso **servidor** ofrece una funcionalidad y uno o más procesos **cliente** la consumen.

Con TCP, el cliente inicia una conexión hacia la dirección y puerto donde escucha el servidor. Con UDP, el cliente envía datagramas al puerto del servidor, sin una conexión TCP previa.

Cliente y servidor pueden ejecutarse en computadores distintos o en la misma máquina mediante `localhost`.
