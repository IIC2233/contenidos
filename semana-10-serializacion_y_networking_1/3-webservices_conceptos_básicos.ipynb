## Servicios web

Un **servicio web** permite que programas intercambien información mediante tecnologías de la web, habitualmente HTTP.

- Un **sitio o aplicación web** suele entregar HTML, CSS y JavaScript para una persona que utiliza un navegador.
- Una **API web** expone operaciones y datos para otras aplicaciones, frecuentemente mediante JSON.
- Un mismo servidor puede ofrecer páginas web y endpoints de una API.

Cliente y servidor deben conocer el protocolo, el formato de los datos y el significado de cada operación.

<a id="api"></a>
## API (*Application Programming Interface*)

Una **API** define cómo otros programas interactúan con un componente o servicio. Puede estar formada por funciones, métodos o endpoints web.

Una API web documenta su URL base, endpoints, métodos HTTP, parámetros, cuerpos, autenticación, respuestas y códigos de estado.

Un navegador es un cliente HTTP, pero no todo servidor visitado es “una API”: puede solicitar documentos HTML y, además, ejecutar código que consume APIs.

<a id="url"></a>
## URL (*Uniform Resource Locator*)

Una **URL** indica cómo localizar un recurso. Distintas URLs pueden conducir al mismo recurso.

### Estructura de una URL

```text
https://api.ejemplo.com:8080/users/42/posts?page=2&limit=10#comentarios
└──┬──┘ └──────┬───────┘└─┬─┘└─────┬──────┘└──────┬──────┘└────┬─────┘
 esquema      host       puerto    ruta         query string    fragmento
```

| Componente | Ejemplo | Descripción |
|---|---|---|
| Esquema | `https` | Mecanismo de acceso. |
| Host | `api.ejemplo.com` | Dominio o dirección IP. |
| Puerto | `8080` | Puerto opcional; HTTP usa 80 y HTTPS 443 por defecto. |
| Ruta | `/users/42/posts` | Recurso dentro del servicio. |
| *Query string* | `page=2&limit=10` | Parámetros después de `?`. |
| Fragmento | `#comentarios` | Sección del documento; no se envía al servidor. |

<a id="http"></a>
## HTTP (*Hypertext Transfer Protocol*)

HTTP es una familia de protocolos de solicitud y respuesta. Es **stateless**: cada solicitud tiene significado por sí misma. La aplicación sí puede guardar usuarios, compras y sesiones.

### Métodos HTTP

La semántica principal está en [RFC 9110](https://www.rfc-editor.org/rfc/rfc9110.html); `PATCH` se define en RFC 5789.

- `GET` solicita una representación y debe ser seguro.
- `HEAD` es como `GET`, pero sin contenido de respuesta.
- `POST` pide procesar el contenido; frecuentemente crea recursos, pero no se limita a eso.
- `PUT` crea o reemplaza el estado del recurso identificado.
- `PATCH` aplica cambios parciales.
- `DELETE` solicita eliminar la asociación con el recurso.
- `OPTIONS` consulta opciones; `CONNECT` crea un túnel y `TRACE` realiza diagnóstico.

### Códigos de estado

| Rango | Categoría |
|---|---|
| 1xx | Informacional |
| 2xx | Éxito |
| 3xx | Redirección |
| 4xx | Error del cliente |
| 5xx | Error del servidor |

Códigos frecuentes: `200 OK`, `201 Created`, `202 Accepted`, `204 No Content`, `304 Not Modified`, `400 Bad Request`, `401 Unauthorized`, `403 Forbidden`, `404 Not Found`, `405 Method Not Allowed`, `409 Conflict`, `415 Unsupported Media Type`, `429 Too Many Requests`, `500 Internal Server Error` y `503 Service Unavailable`.

### Mensajes HTTP

Solicitudes y respuestas contienen una línea inicial, campos de encabezado y contenido opcional. `Content-Type` describe el formato, por ejemplo `application/json`, `text/html; charset=utf-8` o `image/png`.

Referencias: [MDN](https://developer.mozilla.org/docs/Web/HTTP), [IANA](https://www.iana.org/assignments/http-status-codes/) y [RFC 9110](https://www.rfc-editor.org/rfc/rfc9110.html).

<a id="tls"></a>
## Bonus: TLS (*Transport Layer Security*)

TLS crea un canal seguro. En `https://`, HTTP se comunica mediante TLS. HTTP/1.1 y HTTP/2 normalmente usan TLS sobre TCP; HTTP/3 integra TLS 1.3 dentro de QUIC sobre UDP.

TLS proporciona confidencialidad, integridad y autenticación del servidor. La autenticación del cliente es opcional.

### *Handshake* TLS 1.3 simplificado

1. El cliente ofrece versiones, algoritmos, grupos y una participación de clave.
2. El servidor selecciona parámetros y responde con su participación.
3. Ambos derivan secretos, normalmente mediante `(EC)DHE`, o usan una clave precompartida.
4. Con certificados, `Certificate` y `CertificateVerify` autentican al servidor.
5. `Finished` confirma las claves y la negociación.
6. Comienza el intercambio cifrado y autenticado.

El intercambio `(EC)DHE` establece claves; el certificado y la firma autentican. SSL es el antecesor obsoleto de TLS.

Un certificado se valida por su cadena de confianza, vigencia, dominio y firmas. La interfaz visual del navegador cambia, por lo que no debe dependerse de un icono específico.

| Característica | HTTP sin TLS | HTTPS |
|---|---|---|
| Cifrado | No | Sí |
| Puerto habitual | 80 | 443 |
| Autenticación del servidor | No | Sí |
| Uso recomendado | Ejercicios locales | Producción y datos reales |

TLS agrega trabajo criptográfico, pero HTTP sin TLS no siempre es más rápido: también influyen reutilización de conexiones y HTTP/2 o HTTP/3.
